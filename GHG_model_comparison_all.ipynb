# Comparison of all statistical models for predicting N<sub>2</sub>O and CH<sub>4</sub> flux over Ränskälänkorpi clearcut in 2022

Models are

**Simple model**
$$
\ln{F_{i,k}} = \alpha_k + \beta_k \frac{T_i - 10^{\circ} \text{C}}{10^{\circ} \text{C}} \\
i = 1 \ldots \text{\# observations} \\
k \in \left[  \mathrm{N_2O} ; \mathrm{CH_4} \right] \\
T_i = \text{ Observed air temperature (K)}
$$

**Simple model with soil moisture**
$$
\ln{F_{i,k}} = \alpha_k + \beta_k \frac{T_i - 10^{\circ} \text{C}}{10^{\circ} \text{C}} + \zeta \theta \\
$$
**Full model**
$$
\ln{F_{i,k}} = \alpha_k + \beta_k f(T_{i}) + \sum_{j=1}^{N} \gamma_{j,k}x_{i,j} + \delta_{j,k} x_{i,j} f(T_{i}) \\

f(T_{i}) = \frac{T_i - 10^{\circ} \text{C}}{10^{\circ} \text{C}}
$$

**Full model without ST specific T dependency**
$$
\ln{F_{i,k}} = \alpha_k + \beta_k f(T_{i}) + \sum_{j=1}^{N} \gamma_{j,k}x_{i,j}\\
$$

**Full model with soil moisture**
$$
\ln{F_{i,k}} = \alpha_k + \beta_k f(T_{i}) +\zeta \theta + \sum_{j=1}^{N} \gamma_{j,k}x_{i,j} + \delta_{j,k} x_{i,j} f(T_{i}) \\
$$

**Full model with soil moisture and without ST specific dependency**
$$
\ln{F_{i,k}} = \alpha_k + \beta_k f(T_{i}) +\zeta \theta + \sum_{j=1}^{N} \gamma_{j,k}x_{i,j} \\
$$

Full models are run with 9, 6, 5, 4 and 3 surface types and labelled with "ST9/6/5/4/3" correspondingly

## Compute log-likelihood

```python
for i,model in enumerate(models_n2o):
    print(model_names[i])
    with model["model"]:
        pm.compute_log_likelihood(model["model_res"])
```

## Create dictionary of model traces

```python
traces = [model['model_res'] for model in models_n2o]
models_dict = {model_name: trace for (model_name, trace) in zip(model_names, traces)}
```

```python
df_comp_loo = az.compare(models_dict)
df_comp_loo
```
